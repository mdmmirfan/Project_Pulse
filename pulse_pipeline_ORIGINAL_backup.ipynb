```python
# Cell 2: Download Multiple Playlists (JS Solver + Color Logging)
import os
import yt_dlp

# Custom aesthetic loggers for VS Code
def print_error(msg):
    # \033[91m is the ANSI code for Red
    print(f"\033[91m🚨 ERROR: {msg}\033[0m")

def print_success(msg):
    # \033[92m is the ANSI code for Green
    print(f"\033[92m✅ SUCCESS: {msg}\033[0m")

def download_playlist_audio(playlist_url, output_folder):
    os.makedirs(output_folder, exist_ok=True)
    
    ydl_opts = {
        'format': 'm4a/bestaudio/best', 
        'outtmpl': f'{output_folder}/%(title)s.%(ext)s',
        'postprocessors': [{
            'key': 'FFmpegExtractAudio',
            'preferredcodec': 'wav',
            'preferredquality': '192',
        }],
        'quiet': False, 
        'extract_flat': False,
        'cookiefile': 'cookies.txt', 
        'extractor_args': {
            'youtube': ['player_client=android'] 
            
        }
    }
    
    print(f"\n⏳ Starting extraction for {output_folder}...")
    try:
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            ydl.download([playlist_url])
        print_success(f"Finished downloading to {output_folder}.")
    except Exception as e:
        print_error(f"Extraction failed for {output_folder}.")
        print_error(str(e))

# --- YOUR LINKS ---
playlists_to_mine = {
    "https://www.youtube.com/playlist?list=PL0fY2f8HUipy_ou5UTNWUEU_Pxzf9mt2B": "data/cluster_a_love",
    "https://www.youtube.com/watch?v=kl8T6tsOZJk&list=PL0fY2f8HUipwm3fyaJHgvLkGKq9llc13j": "data/cluster_b_dislike"
}

for url, folder in playlists_to_mine.items():
    download_playlist_audio(url, output_folder=folder)
```

```python
# Cell 3: Extracting Sonic Features (MFCCs)
import librosa
import numpy as np
import pandas as pd
import os
import warnings

warnings.filterwarnings('ignore')

def process_audio_folder(folder_path, label, duration=30, offset=30):
    """
    Scans a folder of .wav files, extracts MFCCs, and returns a list of dictionaries.
    """
    features_list = []
    
    print(f"\n🔍 Scanning {folder_path} for sonic textures...")
    
    # Iterate through every file in the folder
    for filename in os.listdir(folder_path):
        if filename.endswith(".wav"):
            file_path = os.path.join(folder_path, filename)
            try:
                # Load a 30-second chunk, skipping the first 30 seconds
                y, sr = librosa.load(file_path, sr=22050, offset=offset, duration=duration)
                
                # Extract 13 MFCCs (standard for timbral analysis)
                mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
                
                # Average the MFCCs across time to get a single 13-value fingerprint
                mfccs_mean = np.mean(mfccs.T, axis=0)
                
                # Store the data
                feature_dict = {
                    'track_name': filename.replace('.wav', ''),
                    'label': label
                }
                
                # Add the 13 MFCC values as separate columns
                for i in range(13):
                    feature_dict[f'mfcc_{i+1}'] = mfccs_mean[i]
                    
                features_list.append(feature_dict)
                print(f"  ✓ Processed: {filename[:30]}...") 
                
            except Exception as e:
                print(f"  ✗ Failed to process {filename}: {e}")
                
    return features_list

# --- RUN THE EXTRACTION ---
# Process both clusters
cluster_a_features = process_audio_folder("data/cluster_a_love", label="Like")
cluster_b_features = process_audio_folder("data/cluster_b_dislike", label="Dislike")

# Combine the two lists into one big dataset
all_features = cluster_a_features + cluster_b_features

# Convert to a Pandas DataFrame for easy analysis
df_pulse = pd.DataFrame(all_features)

print(f"\n🎉 Extraction Complete! Extracted features for {len(df_pulse)} total tracks.")

# Display the first 5 rows to ensure it looks clean
df_pulse.head()
```

```python
# Cell 5: Upgraded Feature Extraction (Texture + Structure)

warnings.filterwarnings('ignore')

def extract_advanced_features(folder_path, label, duration=30, offset=30):
    features_list = []
    print(f"🔍 Scanning {folder_path} for advanced acoustics...")
    
    for filename in os.listdir(folder_path):
        if filename.endswith(".wav"):
            file_path = os.path.join(folder_path, filename)
            try:
                y, sr = librosa.load(file_path, sr=22050, offset=offset, duration=duration)
                
                # 1. MFCCs (The 13 Texture Layers)
                mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
                mfccs_mean = np.mean(mfccs.T, axis=0)
                
                # 2. Spectral Centroid (Brightness/Center of Mass)
                centroid = librosa.feature.spectral_centroid(y=y, sr=sr)
                centroid_mean = np.mean(centroid)
                
                # 3. Zero-Crossing Rate (Percussiveness/Noise)
                zcr = librosa.feature.zero_crossing_rate(y)
                zcr_mean = np.mean(zcr)
                
                # 4. Tempo (BPM)
                tempo, _ = librosa.beat.beat_track(y=y, sr=sr)
                tempo_val = float(tempo[0]) if isinstance(tempo, np.ndarray) else float(tempo)
                
                # Compile the dictionary
                feature_dict = {
                    'track_name': filename.replace('.wav', ''),
                    'label': label,
                    'brightness_centroid': centroid_mean,
                    'percussiveness_zcr': zcr_mean,
                    'tempo_bpm': tempo_val
                }
                
                for i in range(13):
                    feature_dict[f'mfcc_{i+1}'] = mfccs_mean[i]
                    
                features_list.append(feature_dict)
                
            except Exception as e:
                print(f"  ✗ Failed: {filename}")
                
    return features_list

# Run the extraction on the files you already have
cluster_a_adv = extract_advanced_features("data/cluster_a_love", label="Like")
cluster_b_adv = extract_advanced_features("data/cluster_b_dislike", label="Dislike")

# Create the new Master DataFrame
df_advanced = pd.DataFrame(cluster_a_adv + cluster_b_adv)
print(f"✅ Advanced Extraction Complete! Processed {len(df_advanced)} tracks.")
```

```python
# Cell 9: The t-SNE Clustering Algorithm
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler

plt.style.use('dark_background')

# 1. Gather all 16 features (13 MFCCs + Brightness + Percussiveness + Tempo)
# We drop 'track_name' and 'label' because the math engine only reads numbers
feature_cols = [col for col in df_advanced.columns if col not in ['track_name', 'label']]
X_raw = df_advanced[feature_cols]

# 2. Scale the data 
# (Crucial step: Tempo is in the 100s, but Percussiveness is 0.05. 
# Scaling makes them equal weight to the machine).
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)

# 3. Initialize and run t-SNE
# perplexity controls how the algorithm balances local vs global aspects of your data
# 30 is a great default, but we can tweak it if the clusters are messy
tsne = TSNE(n_components=2, perplexity=25, random_state=42, init='pca', learning_rate='auto')
tsne_results = tsne.fit_transform(X_scaled)

# 4. Add the new 2D coordinates back to our dataframe
df_advanced['tsne_x'] = tsne_results[:, 0]
df_advanced['tsne_y'] = tsne_results[:, 1]

# 5. Plot the final separated clusters
plt.figure(figsize=(10, 8))
sns.scatterplot(
    x='tsne_x', y='tsne_y',
    hue='label',
    palette=["#00ffcc", "#ff3366"], # Cyan (Like), Pink (Dislike)
    data=df_advanced,
    s=120,
    alpha=0.9,
    edgecolor='white'
)

plt.title("Acoustic DNA Clusters (t-SNE Projection)", fontsize=16, pad=15)
plt.xlabel("Sonic Dimension 1")
plt.ylabel("Sonic Dimension 2")
plt.legend(title='Preference', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
# Cell 10: Isolating Your Acoustic Extremes
import pandas as pd

# 1. Isolate ONLY the songs you love
df_likes = df_advanced[df_advanced['label'] == 'Like'].copy()

# 2. Calculate "Micro-Texture Complexity" 
# (How much the tiny MFCC bars jump around, indicating intricate layering)
mfcc_cols = [f'mfcc_{i}' for i in range(3, 14)]
df_likes['texture_complexity'] = df_likes[mfcc_cols].var(axis=1)

# 3. Sort and pull the top 5 for each category
warmest_tracks = df_likes.sort_values(by='brightness_centroid', ascending=True).head(5)
smoothest_tracks = df_likes.sort_values(by='percussiveness_zcr', ascending=True).head(5)
complex_tracks = df_likes.sort_values(by='texture_complexity', ascending=False).head(5)

# --- Output the Results ---
print("🔥 THE WARMEST TRACKS (Lowest Spectral Centroid)")
print("-" * 50)
for index, row in warmest_tracks.iterrows():
    print(f"🎵 {row['track_name']} (Score: {row['brightness_centroid']:.0f})")

print("\n🌊 THE SMOOTHEST TRACKS (Lowest Percussiveness)")
print("-" * 50)
for index, row in smoothest_tracks.iterrows():
    print(f"🎵 {row['track_name']} (Score: {row['percussiveness_zcr']:.4f})")

print("\n🎼 THE MOST COMPLEX TEXTURES (Highest MFCC Variance)")
print("-" * 50)
for index, row in complex_tracks.iterrows():
    print(f"🎵 {row['track_name']} (Score: {row['texture_complexity']:.2f})")
```

```python
# Cell 11: Isolating Your Acoustic Dealbreakers (Dislikes)
import pandas as pd

# 1. Isolate ONLY the songs you dislike
df_dislikes = df_advanced[df_advanced['label'] == 'Dislike'].copy()

# 2. Calculate "Micro-Texture Complexity" (Variance of MFCCs 3-13)
mfcc_cols = [f'mfcc_{i}' for i in range(3, 14)]
df_dislikes['texture_complexity'] = df_dislikes[mfcc_cols].var(axis=1)

# 3. Sort and pull the top 5 for the exact OPPOSITE extremes
brightest_tracks = df_dislikes.sort_values(by='brightness_centroid', ascending=False).head(5)
chaotic_tracks = df_dislikes.sort_values(by='percussiveness_zcr', ascending=False).head(5)
flat_tracks = df_dislikes.sort_values(by='texture_complexity', ascending=True).head(5)

# --- Output the Results ---
print("🚨 THE PIERCING TRACKS (Highest Spectral Centroid / Brightness)")
print("-" * 50)
for index, row in brightest_tracks.iterrows():
    print(f"🎵 {row['track_name']} (Score: {row['brightness_centroid']:.0f})")

print("\n🥁 THE CHAOTIC TRACKS (Highest Percussiveness / ZCR)")
print("-" * 50)
for index, row in chaotic_tracks.iterrows():
    print(f"🎵 {row['track_name']} (Score: {row['percussiveness_zcr']:.4f})")

print("\n🥱 THE FLATTEST TEXTURES (Lowest MFCC Variance / Simplest Layering)")
print("-" * 50)
for index, row in flat_tracks.iterrows():
    print(f"🎵 {row['track_name']} (Score: {row['texture_complexity']:.2f})")
```

```python
# Cell 12: Unsupervised Clustering (The Blind Test)
from sklearn.cluster import KMeans

plt.style.use('dark_background')

# 1. Grab our scaled 16-dimensional data from earlier
# We are explicitly ignoring the 'label' (Like/Dislike)
X_blind = X_scaled 

# 2. Initialize K-Means. We are asking the machine to find 4 distinct groups.
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_blind)

# 3. Add the machine's blind clusters to a COPY of our dataframe
df_unsupervised = df_advanced.copy()
df_unsupervised['blind_cluster'] = clusters

# 4. Visualize the results on our existing t-SNE map
plt.figure(figsize=(10, 8))
sns.scatterplot(
    x='tsne_x', y='tsne_y',
    hue='blind_cluster',
    palette='Set3', # Using a neutral, distinct color palette
    data=df_unsupervised,
    s=120,
    alpha=0.9,
    edgecolor='white'
)

plt.title("Machine's Blind Acoustic Clusters (K-Means)", fontsize=16, pad=15)
plt.xlabel("Sonic Dimension 1")
plt.ylabel("Sonic Dimension 2")
plt.legend(title='Machine Cluster', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.2)
plt.tight_layout()
plt.show()

# 5. Let's peek at the songs inside the machine's new clusters
for i in range(4):
    print(f"\n🎧 CLUSTER {i} 🎧")
    print("-" * 40)
    cluster_songs = df_unsupervised[df_unsupervised['blind_cluster'] == i]
    
    # Print the purity of the cluster
    likes_count = len(cluster_songs[cluster_songs['label'] == 'Like'])
    dislikes_count = len(cluster_songs[cluster_songs['label'] == 'Dislike'])
    print(f"Contains: {likes_count} Likes | {dislikes_count} Dislikes")
    
    # Show up to 5 random songs from this cluster to see what it grouped together
    sample_songs = cluster_songs.sample(n=min(5, len(cluster_songs)), random_state=1)
    for _, row in sample_songs.iterrows():
        print(f"- {row['track_name']} (You originally marked as: {row['label']})")
```

```python
# Structure(Tension and Release Analysis)
def extract_tension_features(y, sr):
    """
    Analyzes the macro-structure of the song to identify dynamic tension and release.
    """
    # Extract RMS (Root Mean Square) energy
    rms = librosa.feature.rms(y=y)[0]
    
    # 1. Variance: High variance = lots of valleys and peaks (good dynamic range)
    rms_variance = np.var(rms)
    
    # 2. Peak-to-Trough Ratio: Compare the loudest 10% to the quietest 10%
    # Added a tiny constant (1e-6) to avoid division by zero errors on quiet tracks
    trough = np.percentile(rms, 10)
    peak = np.percentile(rms, 90)
    pt_ratio = peak / (trough + 1e-6) 
    
    # 3. Energy Gradient: The average rate of change in volume
    # High gradient means abrupt drops/spikes. Low means slow builds or flat noise.
    energy_gradient = np.mean(np.abs(np.diff(rms)))
    
    return {
        "rms_variance": float(rms_variance),
        "peak_to_trough_ratio": float(pt_ratio),
        "energy_gradient": float(energy_gradient)
    }
```

```python
# Groove (Micro-Timing Analysis)
def extract_groove_features(y, sr):
    """
    Measures how 'human' or 'robotic' a beat is by comparing onsets to a rigid grid.
    """
    # 1. Find the exact millisecond when sudden energy bursts (drum hits) happen
    onset_frames = librosa.onset.onset_detect(y=y, sr=sr)
    onset_times = librosa.frames_to_time(onset_frames, sr=sr)
    
    # 2. Generate a perfectly rigid tempo grid
    tempo, beat_frames = librosa.beat.beat_track(y=y, sr=sr)
    beat_times = librosa.frames_to_time(beat_frames, sr=sr)
    
    # Safety check for ambient tracks with no clear beats
    if len(beat_times) == 0 or len(onset_times) == 0:
        return {"avg_micro_timing_deviation": 0.0, "tempo": float(tempo[0])}
        
    # 3. Measure the deviation
    deviations = []
    for onset in onset_times:
        # Calculate absolute difference between the hit and every beat on the grid
        diffs = np.abs(beat_times - onset)
        # Find the distance to the closest beat
        closest_diff = np.min(diffs)
        deviations.append(closest_diff)
        
    # High deviation = laid-back, human pocket. Low deviation = quantized, rigid.
    avg_deviation = np.mean(deviations)
    
    return {
        "tempo": float(tempo[0]),
        "avg_micro_timing_deviation": float(avg_deviation)
    }
```

```python
# Harmony (Chroma Analysis)
def extract_harmonic_features(y, sr):
    """
    Analyzes Chroma features to determine the stability and dominant notes of the track.
    """
    # Extract Chroma using Constant-Q Transform
    chroma = librosa.feature.chroma_cqt(y=y, sr=sr)
    
    # Get the average intensity of each of the 12 pitch classes over the whole song
    chroma_means = np.mean(chroma, axis=1) 
    
    # 1. Chroma Variance: 
    # High variance = The song stays in a clear key/resolves nicely.
    # Low variance = Highly dissonant, noisy, or constantly shifting keys.
    chroma_variance = np.var(chroma_means)
    
    # 2. Dominant Notes: Find the indices of the top 3 most played pitch classes
    dominant_notes = np.argsort(chroma_means)[-3:]
    
    return {
        "chroma_variance": float(chroma_variance),
        "dominant_note_1": int(dominant_notes[2]), # Strongest note
        "dominant_note_2": int(dominant_notes[1]),
        "dominant_note_3": int(dominant_notes[0])
    }
```

```python
# Cell 20: Local Valence Proxy (Replacing Spotify)
import librosa
import numpy as np

def get_local_valence(y, sr):
    """
    Estimates the 'Valence' (positivity) locally using Audio Brightness and Tonality.
    Returns a score roughly between 0.0 (Dark/Noisy) and 1.0 (Bright/Clear).
    """
    try:
        # 1. Brightness (Spectral Centroid) - Higher frequencies sound 'happier' or more energetic
        cent = librosa.feature.spectral_centroid(y=y, sr=sr)
        avg_brightness = np.mean(cent)
        
        # Normalize brightness roughly to a 0-1 scale (assuming typical music sits around 1500-4000 Hz)
        brightness_norm = min(max((avg_brightness - 1000) / 3000, 0.0), 1.0)
        
        # 2. Dissonance (Spectral Flatness) - Noisy/dissonant sounds are less 'happy'
        flatness = librosa.feature.spectral_flatness(y=y)
        avg_flatness = np.mean(flatness) # 0 is pure tone, 1 is white noise
        
        # Simple proxy: Bright and Tonal = Higher Valence
        valence_estimate = brightness_norm * (1.0 - avg_flatness)
        
        return {"valence_proxy": float(valence_estimate)}
    except Exception as e:
        print(f"Valence calculation error: {e}")
        return {"valence_proxy": None}
```

```python
# Cell 21: Lyrical Sentiment Extraction (Multilingual Update)
import lyricsgenius
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
from deep_translator import GoogleTranslator

# Paste your Genius token here
GENIUS_API_TOKEN = 'YOUR_GENIUS_API_TOKEN_REMOVED'

genius = lyricsgenius.Genius(GENIUS_API_TOKEN)
analyzer = SentimentIntensityAnalyzer()
translator = GoogleTranslator(source='auto', target='en')

def get_lyrical_sentiment(track_name, artist_name=""):
    """Scrapes lyrics, auto-translates to English, and scores the sentiment."""
    try:
        song = genius.search_song(track_name, artist_name)
        if song:
            original_lyrics = song.lyrics
            
            # Auto-translate the lyrics to English so VADER can read them
            try:
                english_lyrics = translator.translate(original_lyrics[:4999]) # 5000 char limit
            except Exception as e:
                print(f"Translation failed: {e}")
                english_lyrics = original_lyrics # Fallback to original if it fails
                
            # Score the translated text
            sentiment = analyzer.polarity_scores(english_lyrics)
            return {"lyrical_sentiment": sentiment['compound']}
            
        return {"lyrical_sentiment": None}
    except Exception as e:
        print(f"Genius API Error: {e}")
        return {"lyrical_sentiment": None}
```

```python
# Cell 22: Harmonic-Percussive Source Separation

def extract_stem_features(y):
    """Separates melody from drums and calculates which one dominates the track."""
    # Split the audio wave into two separate waves
    y_harmonic, y_percussive = librosa.effects.hpss(y)
    
    # Calculate the raw physical energy of the melody vs. the drums
    harmonic_energy = np.sum(y_harmonic ** 2)
    percussive_energy = np.sum(y_percussive ** 2)
    
    # Calculate the ratio (High ratio = Melody dominant, Low ratio = Drum dominant)
    melody_to_drum_ratio = harmonic_energy / (percussive_energy + 1e-6)
    
    return {
        "melody_energy": float(harmonic_energy),
        "drum_energy": float(percussive_energy),
        "melody_to_drum_ratio": float(melody_to_drum_ratio)
    }
```

```python
# Cell 23: Deep Neural Audio Embeddings
import torch
from transformers import AutoProcessor, ASTModel

# Load a pre-trained Audio AI model (this will download a large file the first time you run it!)
processor = AutoProcessor.from_pretrained("MIT/ast-finetuned-audioset-10-10-0.4593")
model = ASTModel.from_pretrained("MIT/ast-finetuned-audioset-10-10-0.4593")

def extract_neural_footprint(y, sr):
    """Feeds the audio into a neural network to get an abstract 'footprint' score."""
    # Deep learning models require audio to be at exactly 16,000 Hz
    if sr != 16000:
        y_16k = librosa.resample(y, orig_sr=sr, target_sr=16000)
    else:
        y_16k = y
        
    # Prepare the audio and pass it through the brain
    inputs = processor(y_16k, sampling_rate=16000, return_tensors="pt")
    
    with torch.no_grad():
        outputs = model(**inputs)
        
    # Extract the abstract hidden pattern and average it into a single number
    last_hidden_state = outputs.last_hidden_state
    neural_vibe = torch.mean(last_hidden_state).item()
    
    return {"neural_footprint": neural_vibe}
```

```python
# Cell 28: Building the 3D Neural Map
import plotly.express as px

# 1. Update the AI to output all 768 dimensions instead of averaging them
def extract_full_neural_vector(y, sr):
    if sr != 16000:
        y_16k = librosa.resample(y, orig_sr=sr, target_sr=16000)
    else:
        y_16k = y
        
    inputs = processor(y_16k, sampling_rate=16000, return_tensors="pt")
    with torch.no_grad():
        outputs = model(**inputs)
        
    # Keep the 768 distinct numbers that make up the track's "footprint"
    full_vector = torch.mean(outputs.last_hidden_state, dim=1).squeeze().numpy()
    return full_vector

def process_neural_3d(folder_path, preference_label):
    data = []
    if not os.path.exists(folder_path): return data

    for filename in os.listdir(folder_path):
        if filename.endswith(".wav") or filename.endswith(".mp3"):
            file_path = os.path.join(folder_path, filename)
            track_name = filename.replace('.wav', '').replace('.mp3', '')
            
            print(f"Extracting AI Brain Waves [{preference_label}]: {track_name}...")
            try:
                y, sr = librosa.load(file_path, sr=22050)
                vector = extract_full_neural_vector(y, sr)
                
                # Store the label, the track name, and all 768 numbers
                row = {'Track': track_name, 'Preference': preference_label}
                for i in range(768):
                    row[f'dim_{i}'] = vector[i]
                data.append(row)
            except Exception as e:
                print(f"Error on {filename}: {e}")
    return data

print("Starting 3D Neural Extraction. This will take roughly 15-20 minutes...\n")
liked_neural = process_neural_3d(liked_dir, "Liked")
disliked_neural = process_neural_3d(disliked_dir, "Disliked")

# Combine into a massive dataframe
df_neural = pd.DataFrame(liked_neural + disliked_neural)
print("\n✅ Extraction Complete! Calculating 3D Map...")

# 2. Use t-SNE to compress 768 dimensions down to 3 coordinates
features_only = df_neural.drop(columns=['Track', 'Preference'])

# Adjust perplexity based on how many songs you actually have to avoid errors
perplexity_value = min(5, len(df_neural) - 1) 
tsne = TSNE(n_components=3, random_state=42, perplexity=perplexity_value)
tsne_results = tsne.fit_transform(features_only)

# Add the new 3D coordinates back to our table
df_neural['X'] = tsne_results[:, 0]
df_neural['Y'] = tsne_results[:, 1]
df_neural['Z'] = tsne_results[:, 2]
```
